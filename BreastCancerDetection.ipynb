```python
from tensorflow.keras.models import load_model

# Load the saved model
model = load_model("breast_cancer_detection_model.h5")

# Check model summary
model.summary()
```

```python
import numpy as np
from tensorflow.keras.preprocessing import image

# Load an image for prediction
img_path = "D:/BreastCancerDetectionDataset/test/test12.png"  # Replace with your image path
img = image.load_img(img_path, target_size=(224, 224))  # Adjust size based on your model input
img_array = image.img_to_array(img)
img_array = np.expand_dims(img_array, axis=0)  # Expand dims to fit model input
img_array /= 255.0  # Normalize if needed

# Make prediction
prediction = model.predict(img_array)

# Interpret results (Assuming binary classification: 0 = Benign, 1 = Malignant)
if prediction[0][0] > 0.5:
    print("Malignant (Cancer Detected) 😞")
else:
    print("Benign (No Cancer) 😊")
```
